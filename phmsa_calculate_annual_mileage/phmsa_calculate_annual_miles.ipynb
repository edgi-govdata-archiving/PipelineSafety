```python
import pandas as pd
import os
# need to have openpyxl installed
```

```python
cwd = os.getcwd()
```

```python
####################################
### GAS TRANSMISSION & GATHERING ###
####################################
```

```python
# PHMSA's data collection form changed between 2009-2010 so different sets of column names are needed
# 2001-2009 3. MILES OF PIPE BY DECADE OF INSTALLATION
mi_cols_01_09 = ['RPTID', 'YR', 'OPERATOR_ID', # operator info
                 'B3TON_T', # total for that year
                 'B3TON_1', 'B3TON_2', 'B3TON_3', 'B3TON_4', 'B3TON_5', 'B3TON_6', # totals by decade installed
                 'B3TON_7', 'B3TON_8', 'B3TON_9'] # totals by decade installed
# 2010-present PART J – MILES OF PIPE BY DECADE INSTALLED
mi_cols_10_25 = ["REPORT_NUMBER", "REPORT_YEAR", "OPERATOR_ID", # operator info
                 "PARTJTONTOTAL", # total for that year
                 "PARTJTONUNKWN", "PARTJTONPRE1940", "PARTJTON194049", "PARTJTON195059", "PARTJTON196069", "PARTJTON197079", # totals by decade installed
                 "PARTJTON198089", "PARTJTON199099", "PARTJTON200009", "PARTJTON201019", "PARTJTON202029"] # totals by decade installed
# and a dictionary to map names
col_map = dict(zip(mi_cols_01_09,mi_cols_10_25))
```

```python
# start with Rebecca's snippet for gas transmission & gathering 2010-present
data_path_gt_10_25 = os.path.join(cwd, "data\\annual_gas_transmission_gathering_2010_present")
fl_10_25 = os.listdir(data_path_gt_10_25)
xlsx_fl_10_25 = list(filter(lambda f: f.endswith('.xlsx'), fl_10_25))

gt_ontot_partj_10_25 = pd.DataFrame()

#excl = ('2010.xlsx', '2011.xlsx')
for f in xlsx_fl_10_25:
    #if any(f.endswith(excl) for yrs in excl): continue
    #else:
    data = pd.read_excel(os.path.join(data_path_gt_10_25,f), sheet_name="GT AR Part J", header = 2, usecols=mi_cols_10_25)
    gt_ontot_partj_10_25 = pd.concat([gt_ontot_partj_10_25, data])
```

```python
gt_ontot_partj_10_25
```

```python
# continue with gas transmission & gathering 2001-2009
data_path_gt_01_09 = os.path.join(cwd, "data\\annual_gas_transmission_gathering_2001_2009")
fl_01_09 = os.listdir(data_path_gt_01_09)
xlsx_fl_01_09 = list(filter(lambda f: f.endswith('.xlsx'), fl_01_09))

gt_ontot_partj_01_09 = pd.DataFrame()

for f in xlsx_fl_01_09:
    data = pd.read_excel(os.path.join(data_path_gt_01_09,f), usecols = mi_cols_01_09)
    gt_ontot_partj_01_09 = pd.concat([gt_ontot_partj_01_09, data])
```

```python
# rename columns to match 2010-present data format
gt_ontot_partj_01_09 = gt_ontot_partj_01_09.rename(mapper=col_map, axis=1)
# combine the two timeframes
gt_ontot_partj_01_25 = pd.concat([gt_ontot_partj_01_09, gt_ontot_partj_10_25])
# see the result
gt_ontot_partj_01_25 = gt_ontot_partj_01_25.groupby('REPORT_YEAR').sum()['PARTJTONTOTAL']
gt_ontot_partj_01_25.name = 'gas_transmission'
gt_ontot_partj_01_25
```

```python
########################
### GAS DISTRIBUTION ###
########################
```

```python
# PHMSA's data collection form changed between 2009-2010 so different sets of column names are needed
# 2004-2009 4. MILES OF MAIN AND NUMBER OF SERVICES BY DECADE OF INSTALLATION
mi_cols_04_09 = ['RPTID', 'YR', 'OPERATOR_ID', # operator info
                 'DINSTMT', # total for that year
                 'DINSTM_1', 'DINSTM_2', 'DINSTM_3', 'DINSTM_4', 'DINSTM_5', 'DINSTM_6', 'DINSTM_7', 'DINSTM_8', 'DINSTM_9'] # totals by decade installed
# 2010-present 4. MILES OF MAIN AND NUMBER OF SERVICES BY DECADE OF INSTALLATION
mi_cols_10_25 = ["REPORT_NUMBER", "REPORT_YEAR", "OPERATOR_ID", # operator info
                 "MMILES_BY_DCD_TOTAL", # total for that year
                 "MMILES_BY_DCD_UNK", "MMILES_BY_DCD_PRE1940", "MMILES_BY_DCD_1940_TO_1949", "MMILES_BY_DCD_1950_TO_1959", 
                 "MMILES_BY_DCD_1960_TO_1969", "MMILES_BY_DCD_1970_TO_1979", # totals by decade installed
                 "MMILES_BY_DCD_1980_TO_1989", "MMILES_BY_DCD_1990_TO_1999", "MMILES_BY_DCD_2000_TO_2009", 
                 "MMILES_BY_DCD_2010_TO_2019", "MMILES_BY_DCD_2020_TO_2029"] # totals by decade installed
# and a dictionary to map names
col_map = dict(zip(mi_cols_04_09,mi_cols_10_25))
```

```python
# gas distribution 2004-2009
data_path_gd_04_09 = os.path.join(cwd, "data\\annual_gas_distribution_2004_2009")
fl_04_09 = os.listdir(data_path_gd_04_09)
xlsx_fl_04_09 = list(filter(lambda f: f.endswith('.xlsx'), fl_04_09))

gd_ontot_04_09 = pd.DataFrame()

for f in xlsx_fl_04_09:
    data = pd.read_excel(os.path.join(gd_ontot_04_09,f), usecols = mi_cols_04_09)
    gd_ontot_04_09 = pd.concat([gd_ontot_04_09, data])
```

```python
gd_ontot_04_09
```

```python
# gas distribution 2010-present
data_path_gd_10_25 = os.path.join(cwd, "data\\annual_gas_distribution_2010_present")
fl_10_25 = os.listdir(data_path_gd_10_25)
xlsx_fl_10_25 = list(filter(lambda f: f.endswith('.xlsx'), fl_10_25))

gd_ontot_10_25 = pd.DataFrame()

for f in xlsx_fl_10_25:
    data = pd.read_excel(os.path.join(gd_ontot_10_25,f), header = 2, usecols=mi_cols_10_25)
    gd_ontot_10_25 = pd.concat([gd_ontot_10_25, data])
```

```python
# rename columns to match 2010-present data format
gd_ontot_04_09 = gd_ontot_04_09.rename(mapper=col_map, axis=1)
# combine the two timeframes
gd_ontot_04_25 = pd.concat([gd_ontot_04_09, gd_ontot_10_25])
# see the result
gd_ontot_04_25 = gd_ontot_04_25.groupby('REPORT_YEAR').sum()['MMILES_BY_DCD_TOTAL']
gd_ontot_04_25.name = "gas_distribution"
gd_ontot_04_25
```

```python
########################
### HAZARDOUS LIQUID ###
########################
```

```python
# PHMSA's data collection form changed between 2009-2010 so different sets of column names are needed
# NOTE: hazardous liquids use different time intervals than gas transmission/distribution
# 2004 - 2009
# PART D - MILES OF PIPE BY DECADE INSTALLED
mi_cols_04_09 = ['RPTID', 'YR', 'OPERATOR_ID', # operator info
                 'DINSTMT', # total for that year
                 'DINSTM_1', # Pre-20 or Unknown
                 'DINSTM_2', # 1920-1929
                 'DINSTM_3', # 1930-1939
                 'DINSTM_4', # 1940-1949
                 'DINSTM_5', # 1950-1959
                 'DINSTM_6', # 1960-1969
                 'DINSTM_7', # 1970-1979
                 'DINSTM_8', # 1980-1989
                 'DINSTM_9', # 1990-1999
                 'DINSTM_10'] # 2000-2009
# 2010 - present
# PART I - MILES OF PIPE BY DECADE INSTALLED
mi_cols_10_25 = ["REPORT_NUMBER", "REPORT_YEAR", "OPERATOR_ID", # operator info
                 "PARTITOTAL", # total for that year
                 "PARTIUNKWN", # Unknown
                 'PARTIPRE20', # Pre-20s
                 'PARTI192029', # 1920-1929
                 'PARTI193039', # 1930-1939
                 'PARTI194049', # 1940-1949
                 'PARTI195059', # 1950-1959
                 'PARTI196069', # 1960-1969
                 'PARTI197079', # 1970-1979
                 'PARTI198089', # 1980-1989
                 'PARTI199099', # 1990-1999
                 'PARTI200009', # 2000-2009
                 'PARTI201019', # 2010-2019
                 'PARTI202029'] # 2020-2029
```

```python
# hazardous liquid 2004-2009
data_path_hl_04_09 = os.path.join(cwd, "data\\annual_hazardous_liquid_2004_2009")
fl_04_09 = os.listdir(data_path_hl_04_09)
xlsx_fl_04_09 = list(filter(lambda f: f.endswith('.xlsx'), fl_04_09))

hl_04_09 = pd.DataFrame()

for f in xlsx_fl_04_09:
    data = pd.read_excel(os.path.join(data_path_hl_04_09,f), usecols = mi_cols_04_09)
    hl_04_09 = pd.concat([hl_04_09, data])
```

```python
hl_04_09
```

```python
# hazardous liquid 2010-present
data_path_hl_10_25 = os.path.join(cwd, "data\\annual_hazardous_liquid_2010_present")
fl_10_25 = os.listdir(data_path_hl_10_25)
xlsx_fl_10_25 = list(filter(lambda f: f.endswith('.xlsx'), fl_10_25))

hl_10_25 = pd.DataFrame()

for f in xlsx_fl_10_25:
    data = pd.read_excel(os.path.join(data_path_hl_10_25,f), sheet_name="HL AR Part I", header = 2, usecols=mi_cols_10_25)
    hl_10_25 = pd.concat([hl_10_25, data])
```

```python
hl_10_25
```

```python
# group by year and calculate sum
# 2004-2009
hl_04_09 = hl_04_09.groupby('YR').sum()['DINSTMT']
hl_04_09
```

```python
# group by year and calculate sum
# 2010-2025
hl_10_25 = hl_10_25.groupby('REPORT_YEAR').sum()['PARTITOTAL']
hl_10_25
```

```python
# combine the two timeframes
hl_04_25 = pd.concat([hl_04_09, hl_10_25])
hl_04_25.name = 'hazardous_liquids'
hl_04_25
```
